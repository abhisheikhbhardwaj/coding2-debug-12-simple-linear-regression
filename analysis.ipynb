# Simple Linear Regression

Fit and interpret a one-predictor regression of salary on training hours.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import statsmodels.formula.api as smf
from pathlib import Path

sns.set_theme(style="whitegrid")

# Identify repository root relative to current notebook file
NOTEBOOK_DIR = Path.cwd()
DATA_DIR = NOTEBOOK_DIR / "data"
```

```python
# Corrected filename: career_outcomes.csv
df = pd.read_csv(DATA_DIR / "career_outcomes.csv")
df.head()
```

## Analytic question

How much does salary tend to change with additional training hours? State the outcome, predictor, expected direction, and one plausible omitted variable.

**Outcome:** salary (thousand EUR)
**Predictor:** training hours
**Expected direction:** positive (more training hours, higher salary)
**Plausible omitted variable:** AI tool use. It is correlated with training hours (r ≈ 0.62) and could also affect salary.

```python
fig, ax = plt.subplots(figsize=(7, 4))
sns.scatterplot(data=df, x="training_hours", y="salary_k_eur", alpha=0.65, ax=ax)
sns.regplot(data=df, x="training_hours", y="salary_k_eur", scatter=False, color="black", ax=ax)
ax.set(title="Salary and training hours", xlabel="Training hours", ylabel="Salary, thousand EUR")
```

```python
model = smf.ols("salary_k_eur ~ training_hours", data=df).fit()
print(model.summary())
```

```python
intercept = model.params["Intercept"]
slope = model.params["training_hours"]
training_sd = df["training_hours"].std()
effect_per_sd = slope * training_sd
print(f"Predicted salary with no training: {intercept:.1f} thousand EUR")
print(f"Estimated salary change per additional training hour: {slope:.3f} thousand EUR")
print(f"Estimated salary change per {training_sd:.1f}-hour (one SD) increase: {effect_per_sd:.2f} thousand EUR")
```

```python
new_workers = pd.DataFrame({"training_hours": [10, 40, 80]})
predictions = model.get_prediction(new_workers).summary_frame(alpha=0.05)
new_workers.join(predictions[["mean", "obs_ci_lower", "obs_ci_upper"]])
```

```python
diagnostics = df.assign(fitted=model.fittedvalues, residual=model.resid)
fig, ax = plt.subplots(figsize=(7, 4))
sns.scatterplot(data=diagnostics, x="fitted", y="residual", alpha=0.65, ax=ax)
ax.axhline(0, color="black", linewidth=1)
ax.set(title="Residuals versus fitted values", xlabel="Fitted salary", ylabel="Residual")
```

## Interpretation

State the slope as an association, with units, and name one omitted variable that could confound it.

Each additional training hour is associated with about 0.295 thousand EUR (about €295) higher salary (95% CI 0.18 to 0.41). This is an association, not a causal effect. Training hours explain only about 9% of the variation in salary (R² = 0.092), so individual salaries vary widely around the fitted line, and the prediction intervals are wide (for example, 46 to 105 thousand EUR at 40 hours).

AI tool use could confound the slope: workers who train more also use AI tools more, and AI tool use may itself relate to salary.
